# Sentiment Classifier NLP

We will train our model using LogisticRegression on a dataset to determine if a word is positive or negative. This will be scored by splitting user query input by word and normalized based on ratio between length and score. Then simply set a threshold at which the user query is considered positive or negative!

## Logistic Regression Sentiment Model

### Data Loading/Exploration

#### Load txt files for pos and neg

```python
with open("datasets/pos.txt") as pos_txt:
    pos_raw = pos_txt.readlines()

with open("datasets/neg.txt") as neg_txt:
    neg_raw = neg_txt.readlines()

print(pos_raw)
print(neg_raw)
```

### Data Cleaning/Processing

#### Remove \n from words

```python
pos_words = [word.replace("\n", "") for word in pos_raw]
neg_words = [word.replace("\n", "") for word in neg_raw]

print(pos_words)
print(neg_words)
```

#### Remove empty chars from words

```python
pos_words = [word.replace("", "") for word in pos_words]
neg_words = [word.replace("", "") for word in neg_words]

print(pos_words)
print(neg_words)
```

#### Make dictionary of word as key and sentiment as value

```python
pos_words_dict = {word: 1 for word in pos_words}
neg_words_dict = {word: 0 for word in neg_words}

words_dict = pos_words_dict | neg_words_dict

print(words_dict)    

pos_count = 0 
neg_count = 0

for word, sent in words_dict.items():
    if sent == 0:
        # negative
        neg_count += 1
    else:
        # positive
        pos_count += 1

print("Positive:", pos_count)
print("Negative:", neg_count)
```

#### Create dataframe

```python
import pandas as pd

# convert to list of tuples and split by columns
sent_df = pd.DataFrame(list(words_dict.items()), columns=["word", "sentiment"])
print(sent_df.head())
```

### Model Building

#### Define features and to predict

```python
X = sent_df["word"] # features
y = sent_df["sentiment"] # to predict
```

#### Train test split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, test_size=0.25, stratify=y)
```

#### Vectorize words

```python
# tdidfvectorizer works better with text to give less weight to ("and", "or", etc)
from sklearn.feature_extraction.text import TfidfVectorizer

vec = TfidfVectorizer()

# vectorizer model learning
vec.fit(X_train)

# use trained vectorizer model to transform from text -> vectors with X sets
X_train_vec = vec.transform(X_train)
X_test_vec = vec.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(class_weight="balanced")

model.fit(X_train_vec, y_train)
```

### Model Evaluation

#### Get predictions

```python
y_pred = model.predict(X_test_vec)
```

#### Analyze performance

```python
from sklearn.metrics import accuracy_score, classification_report

scores = classification_report(y_test, y_pred)

print(scores)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
neg = "This product has been nothing but disappointing from the moment I opened the box. The packaging was torn, the instructions were vague, and the build quality felt cheap and fragile. After a few days of use, it started malfunctioning — buttons sticking, random shutdowns, and constant errors that made it nearly impossible to use properly. Customer support was unhelpful and dismissive, offering generic responses that didn’t solve anything. The performance is slow, the battery drains quickly, and the promised features don’t work as advertised. It feels like a waste of money and effort. Every update makes it worse, introducing new bugs instead of fixing old ones. I regret ever buying it and wouldn’t recommend it to anyone. It’s frustrating, overpriced, and completely unreliable. The entire experience has left me angry, dissatisfied, and tired of dealing with this terrible product."
pos = "This product has exceeded every expectation. From the beautifully organized packaging to the smooth setup process, everything about it feels premium and thoughtfully designed. The performance is fast, consistent, and reliable. Every feature works exactly as described, and some even go beyond what I anticipated. The design is sleek, modern, and pleasant to use daily. Battery life lasts surprisingly long, and the interface is intuitive, making it easy for anyone to master quickly. Customer service is prompt, kind, and genuinely helpful, which is refreshing to see these days. I’ve been using it for weeks, and it continues to impress me with its quality and durability. It has improved how I work and made everyday tasks enjoyable. I couldn’t be happier with my purchase."
neutral = "The product arrives in a standard package with average materials. Setup takes a few minutes and follows a basic set of steps as outlined in the included manual. Once powered on, the device performs its advertised functions without noticeable issues. The interface is straightforward, offering common features found in most similar products. Battery life is reasonable but depends on usage frequency and settings. It connects easily to other devices and requires occasional software updates. The construction is neither exceptionally sturdy nor excessively fragile. Customer service responds within a typical timeframe and provides standard assistance. Overall, this product fulfills its core purpose but does not include any outstanding qualities or significant drawbacks. It is functional, adequately built, and represents a typical offering in its category."
```

### Get query

```python
# query = input("Enter your query: ")
query = neutral
```

### Query Processing

```python
def clean_query(query): 
    words = query.lower().split(" ")

    return words

cleaned_query_vec = vec.transform(clean_query(query))

print(cleaned_query_vec)
```

#### Predict query sentiment

```python
pred = model.predict(cleaned_query_vec)

print(pred)
```

#### Evaluate if positive or negative

```python
pos_pred = 0
neg_pred = 0

for p in pred:
    if p == 0:
        neg_pred += 1
    else:
        pos_pred += 1

if pos_pred > neg_pred:
    print("Positive")
elif pos_pred < neg_pred:
    print("Negative")
else:
    print("Neutral")

print("Positive predicted count:", pos_pred)
print("Negative predicted values:", neg_pred)
```
